```python
import sys
sys.path.append('..')

import pandas as pd
from utils.db_utils import write_table, read_table
```

```python
df_final = read_table("""
with total_emp as (
    select 
        year,
        sum(total_emp_graduate) as total_emp_graduate
    from sc_bronze.dosm_graduates_state
    group by year
)

select 
under.year,
sum(under.skill_total) as total_underemp_skill,
sum(under.time_total) as total_underemp_time,
emp.total_emp_graduate,
round((sum(under.skill_total) * 100.0 / nullif(emp.total_emp_graduate,0))::numeric, 2) as underemp_skill_rate_pct,
round((sum(under.time_total) * 100.0 / nullif(emp.total_emp_graduate,0))::numeric, 2) as underemp_time_rate_pct
from sc_bronze.dosm_underemployment as under
inner join total_emp as emp
on under.year = emp.year
group by under.year, emp.total_emp_graduate
order by under.year
""")

df_final.head(10)
```

```python
write_table(df_final, 'sc_silver', 'underemployment_rate')
```
