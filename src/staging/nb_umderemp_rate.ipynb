```python
import sys
sys.path.append('..')

import pandas as pd
from utils.db_utils import write_table, read_table
```

```python
df_final = read_table("""
with total_emp as (
    select 
        year,
        sum(total_emp_graduate) as total_emp_graduate
    from sc_bronze.dosm_graduates_state
    group by year
)

select
    u.year,
    u.underemp_type,
    sum(u.underemp_graduate) as total_underemp,
    e.total_emp_graduate,
    round((sum(u.underemp_graduate) * 100.0 / nullif(e.total_emp_graduate,0))::numeric, 2) as underemp_rate_pct
from sc_bronze.dosm_underemployment u
inner join total_emp e
    on u.year = e.year
group by u.year, u.underemp_type, e.total_emp_graduate
order by u.year, u.underemp_type
""")

df_final.head()
```

```python
write_table(df_final, 'sc_silver', 'underemployment_rate')
```
