```python
import sys
sys.path.append('..')

import pandas as pd
from utils.db_utils import write_table, read_table
```

```python
POP_URL_DATA = 'https://storage.dosm.gov.my/population/population_state.parquet'
pop_df = pd.read_parquet(POP_URL_DATA)
```

```python
pop_df["date"] = pd.to_datetime(pop_df["date"])

pop_df = (
    pop_df[
        (pop_df["date"] >= "2016-01-01") &
        (pop_df["sex"] == "both") &
        (pop_df["age"] == "overall") &
        (pop_df["ethnicity"] == "overall")
    ]
    .copy()
)

pop_df["population"] = (
        pop_df["population"]
        .astype(str)
        .str.replace(",", "", regex=False)
        .astype(float) * 1000
    )

pop_df = pop_df.drop(columns=["sex", "age", "ethnicity"])
```

```python
write_table(pop_df, "sc_bronze", "datagov_population")
```
