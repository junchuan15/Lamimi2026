```python
import sys
sys.path.append('..')

import pandas as pd
from utils.db_utils import write_table, read_table
```

```python
df = pd.read_csv("../../data/Employed Graduates by Economic Activity.csv")
```

```python
df_long = df.melt(
    id_vars="Economic Activity",
    var_name="year",
    value_name="emp_graduates"
)
```

```python
df_long["emp_graduates"] = (
    df_long["emp_graduates"]
    .astype(str)
    .str.replace(",", "")
    .astype(float)
)

df_long["total_emp_graduate"] = df_long["emp_graduates"] * 1000
```

```python
main_sectors = [
    "Agriculture",
    "Mining and Quarrying",
    "Manufacturing",
    "Construction",
    "Services"
]

manufacturing_sub = [
    "Vegetable and animal oils & fats and food processing",
    "Beverages and tobacco products",
    "Textiles, wearing apparel and leather products",
    "Wood products, furniture, paper products and printing",
    "Petroleum, chemical, rubber and plastic products",
    "Non-metallic mineral products, basic metal and fabricated metal products",
    "Electrical, electronic and optical products",
    "Transport equipment, other manufacturing and repair"
]

services_sub = [
    "Utilities",
    "Wholesale and retail trade",
    "Food & beverages and accommodation",
    "Transportation and storage",
    "Information and communication",
    "Finance and insurance",
    "Real estate and business services",
    "Other services"
]
```

```python
df_long["subsector"] = df_long["Economic Activity"]

df_long["sector"] = df_long["Economic Activity"].apply(
    lambda x: "Manufacturing" if x in manufacturing_sub
    else "Services" if x in services_sub
    else x
)

# can remove if want retain subsector
df_long = (
    df_long[
        (df_long["sector"] == df_long["subsector"])
    ]
    .copy()
)
```

```python
# df_final = df_long[["year", "sector", "subsector", "total_emp_graduate"]]
df_final = df_long[["year", "sector", "total_emp_graduate"]]
df_final.head(20)
```
