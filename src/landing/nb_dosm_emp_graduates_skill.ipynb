```python
import sys
sys.path.append('..')

import pandas as pd
from utils.db_utils import write_table, read_table
```

```python
df = pd.read_csv("../../data/Employed Graduates by skill level.csv")
```

```python
df_long = df.melt(
    id_vars="skill_level",
    var_name="year",
    value_name="emp_graduates"
)
```

```python
df_long["emp_graduates"] = (
    df_long["emp_graduates"]
    .astype(str)
    .str.replace(",", "")
    .astype(float)
)

df_long["total_emp_graduate"] = df_long["emp_graduates"] * 1000
```

```python
df_final = df_long[["year", "skill_level", "total_emp_graduate"]]
df_final.head(20)
```

```python
write_table(df_final, "sc_bronze", "dosm_emp_graduates_skill")
```
