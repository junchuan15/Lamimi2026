```python
import sys
sys.path.append('..')

import pandas as pd
from utils.db_utils import write_table
```

```python
df = pd.read_csv("../../data/Mean Salary by State.csv")
```

```python
df_final = df.melt(
    id_vars="state",
    var_name="year",
    value_name="mean_salary_rm"
)

df_final = df_final.sort_values(["state", "year"]).reset_index(drop=True)

df_final.head()
```

```python
write_table(df_final, 'sc_bronze', 'dosm_mean_salary_state')
```
