```python
import sys
sys.path.append('../..')

import pandas as pd
from utils.db_utils import write_table, read_table
```

```python
df = read_table(
    """
    select distinct year::int as year
    from (
        select year::int as year from sc_bronze.dosm_graduates_age
        union
        select EXTRACT(YEAR FROM date)::int as year from sc_bronze.datagov_gdp
        union
        select year::int as year from sc_bronze.dosm_emp_graduates_sector
        union
        select EXTRACT(YEAR FROM date)::int as year from sc_bronze.dosm_jobdemand
    ) t
    order by year
    """
)

df.head()
```

```python
df = df.sort_values("year").reset_index(drop=True)
df["year_id"] = ["Y" + str(i+1).zfill(3) for i in range(len(df))]
df = df[["year_id"] + [c for c in df.columns if c != "year_id"]]
df_final = df.rename(columns={"year": "Year"})
df_final
```

```python
write_table(df_final, "sc_gold", "dim_year")
```
