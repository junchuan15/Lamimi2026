```python
import sys
sys.path.append('../..')

import pandas as pd
from utils.db_utils import write_table, read_table
```

```python
date = read_table("select * from sc_gold.dim_date")
```

```python
df = read_table("select * from sc_bronze.dosm_olf_reason")
df 
```

```python
df["date"] = pd.to_datetime(df["year"], format="%Y")
df = df.merge(
    date[["date", "date_id"]],
    on="date",
    how="left"
)

df_final = df.drop(columns=["year", "date"])
id_cols = ["date_id"]
df_final = df_final[id_cols + [col for col in df_final.columns if col not in id_cols]]
```

```python
df_final["olf_id"] = ["OLF" + str(i+1).zfill(4) for i in range(len(df_final))]
df_final = df_final[["olf_id"] + [c for c in df_final.columns if c != "olf_id"]]
df_final
```

```python
write_table(df_final, "sc_gold", "fact_olf")
```
